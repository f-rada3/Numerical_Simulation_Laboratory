In Statistical Mechanics the one-dimensional (1D) Ising Model is exactly resoluble. The Hamiltonian for a system of N spins with nearest neighbors interaction is given by (in the following I will assume units such that $\mu_B=1$ and $k_B=1$):

$$
H = 
-J\sum_{i=1}^N s_i s_{i+1}
-\frac{h}{2}\sum_{i=1}^N (s_i + s_{i+1}) 
$$

where $\{ s_i \}_{i=1,N}$ are spin variables taking values in $\{-1;1\}$, $J>0$ (ferromagnetic model),and with the assumption $s_{N+1} = s_{1}$.

The partition function for $N$ spins, $Z$, can be exactly computed and it turns out to be:

$$
Z=\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \lambda_1^N + \lambda_2^N
$$

where
$\lambda_{1,2} = 
e^{\beta J} 
\cosh (\beta h) \pm \left[ e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J) \right]^{1/2}$, 
and where $\beta=\frac{1}{k_b T} = \frac{1}{T}$.
Once the partition function has been computed, the Helmholtz free energy is readly obtained as $A = -k_B T \ln Z$. All the thermodynamic quantities of interest can be deduced from $A$ or, equivalently, from $Z$:

- the **Internal Energy** per spin (with $h=0$):

$$
\frac{U(N,T)}{N} = \frac{\langle H \rangle}{N}
= \frac{1}{N} \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}
= -\frac{1}{N}\frac{\partial \ln Z}{\partial \beta}
\underset{h=0}= -J \tanh (\beta J) \Biggl( \frac{1 + \tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

- the **Heat Capacity** (with $h=0$):

$$
C(N,T) = \frac{\partial U(N,T)}{\partial T}
= \frac{\partial \beta}{\partial T}\frac{\partial U}{\partial \beta}
= -k_B\beta^2 \frac{\partial}{\partial \beta} \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z} \right] = \\
= -k_B\beta^2 \left[ -\sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \sum_{\{ s_i \}} H \exp \left( -\beta H\right)\frac{\partial Z^{-1}}{\partial \beta} \right] = \\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]\frac{1}{Z}\frac{\partial Z}{\partial \beta} \right\} =\\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} - \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]^2 \right\} =
k_B\beta^2 \left(\langle H^2 \rangle -\langle H \rangle^2 \right)
$$

$$
\frac{C(N,T)}{N} = \frac{1}{N}\frac{\partial U(N,T)}{\partial T} =\\
\underset{h=0}= 
k_B (\beta J)^2 \left\{ 1+\tanh^N (\beta J)+(N-1)\tanh^2 (\beta J)+(N-1)\frac{\tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)}
- N \left[ \frac{\tanh (\beta J) + \tanh^{N-1} (\beta J)}{1+\tanh^N (\beta J)} \right]^2 \right\}
$$

- The **magnetization** (with $h \not= 0$):

Note that with $h=0$ the magnetization $M(N,T,h=0)=\langle \sum_{i=1,N}s_i \rangle =0$ because in 1D we cannot have a (ferromagnetic) phase transition in absence of an external magnetic field.

$$
M(N,T,h) = -\frac{\partial A}{\partial h} = k_B T \frac{\partial \ln Z}{\partial h} =
\frac{1}{\beta} \frac{1}{Z}\frac{\partial}{\partial h}\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} = \left\langle \sum_{i=1,N}s_i \right\rangle
$$

The exact formula from $M(N,T,h) = k_B T \dfrac{\partial \ln Z}{\partial h}$ is:
$$
\frac{M(N,T,h)}{N}=\frac{e^{\beta J}\; \sinh(\beta h)}{Z} \; \Bigl[\lambda_1^{N-1}\cdot \Bigl(1+\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr) + \lambda_2^{N-1}\cdot \Bigl(1-\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr)\Bigr]
$$

- The **magnetic susceptibility** (with $h=0$):

$$
\chi(N,T) = \frac{\partial M}{\partial h} = 
\frac{\partial}{\partial h}\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} =\\
= \beta\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right]^2 \frac{\exp \left( -\beta H\right)}{Z}
-\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z}
\frac{1}{Z}\frac{\partial Z}{\partial h} = 
\beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle - \left\langle \sum_{i=1,N}s_i \right\rangle^2 \right] = \\
\underset{h=0}= \beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle \right] = \beta e^{2\beta J} \Biggl( \frac{1-\tanh^N (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

### Implementation
L'obiettivo dell'esercizio è simulare un modello di Ising 1D utilizzando due algoritmi di campionamento: Metropolis e Gibbs. Il programma calcola le grandezze termodinamiche ( $U(N,T)$ ), ( $C(N,T)$ ), ( $\chi(N,T)$ ), e ( $M(N,T,h)$ ) come funzione della temperatura ( $T$ ) e confronta i risultati con le curve teoriche.

#### Modello di Ising 1D
Il modello di Ising 1D è descritto dall'Hamiltoniana 
$$H = -J \sum_{i=1}^N s_i s_{i+1} - \frac{h}{2} \sum_{i=1}^N (s_i + s_{i+1}) \, ,$$ 

dove:

- ( $s_i \in {-1, +1}$ ) sono le variabili di spin.
- ( $J > 0$ ) è la costante di accoppiamento (modello ferromagnetico).
- ( $h$ ) è il campo magnetico esterno.
- ( $N = 50$ ) è il numero di spin.
- Si assume ( $s_{N+1} = s_1 $) (condizioni periodiche).

#### Algoritmi di Campionamento

1. **Metropolis Algorithm:**
Per ogni passo Monte Carlo:
- Si sceglie uno spin ( $s_i$ ) casualmente.
- Si calcola la variazione di energia ($ \Delta E $) se lo spin ( $s_i$ ) viene invertito.
- Si accetta la nuova configurazione con probabilità:  
$$A = \min(1, e^{-\beta \Delta E}) \, ,$$
dove  $\beta = 1/T $.

2. **Gibbs Sampling:**
Per ogni passo Monte Carlo:
- Si calcola il campo locale ( $h_{\text{loc}}$ ) percepito dallo spin ( $s_i$ ) dai suoi vicini e dal campo esterno $$h_{\text{loc}} = J (s_{i-1} + s_{i+1}) + h\, .$$
- Lo spin ( $s_i$ ) viene aggiornato direttamente con probabilità condizionale:  $$P(s_i = +1) = \frac{1}{1 + e^{-2\beta h_{\text{loc}}}}, \quad P(s_i = -1) = 1 - P(s_i = +1)\, .$$ 
- Non è necessario un test di accettazione, poiché il Gibbs sampling garantisce che ogni passo rispetti la distribuzione di Boltzmann.

#### Proprietà Calcolate
1. Energia interna per spin:  $U(N,T) = \langle H \rangle _{h=0}\, .$ 

2. Calore specifico:  $C(N,T) = k_B \beta^2 \left( \langle H^2 \rangle _{h=0} - \langle H \rangle _{h=0}^2 \right)\, . $

3. Suscettibilità magnetica: $\chi(N,T) = \beta \langle \left( \sum_{i=1}^N s_i \right)^2 \rangle _{h=0}\, .$ 

4. Magnetizzazione: $ M(N,T,h=0.02) = \langle \sum_{i=1}^N s_i \rangle _{h=0.02}\, . $

```python
# Parameters for theoretical curves
points=100
T = np.linspace(0.4,2.1,num=points)
b = 1/T
h = 0.02
J = 1.0
Ns = 50
th = np.tanh(J/T)
thN= th**Ns
ch = 1/th

# Directory di base per i dati
base_dir_gibbs = "./ex06.1/OUTPUT_gibbs"
base_dir_metropolis = "./ex06.1/OUTPUT_metropolis"

# Temperature da analizzare
temperatures = np.arange(0.50, 2.01, 0.05)  # Da 0.5 a 2.0 con step di 0.05

# Funzione per leggere i dati da una directory specifica
def read_data(base_dir, measure, temp, h_field):
    file_path = os.path.join(base_dir, f"T_{temp:.2f}", h_field, f"{measure}.dat")
    try:
        data = np.loadtxt(file_path, skiprows=1)  # Salta la prima riga (header)
        return data
    except OSError:
        print(f"Errore: File {file_path} non trovato.")
        return None

# Inizializza i dati per le misure
measures = ["total_energy", "specific_heat", "susceptibility", "magnetization"]
data_metropolis = {measure: [] for measure in measures}
data_gibbs = {measure: [] for measure in measures}

# Leggi i dati per ogni temperatura e campo magnetico
for temp in temperatures:
    # h = 0: Energia totale, calore specifico, suscettibilità
    for measure in ["total_energy", "specific_heat", "susceptibility"]:
        # Metropolis
        data_m = read_data(base_dir_metropolis, measure, temp, "h0.00_thermodynamics")
        if data_m is not None:
            data_metropolis[measure].append((temp, data_m[:, 2], data_m[:, 3]))  # Media e errore

        # Gibbs
        data_g = read_data(base_dir_gibbs, measure, temp, "h0.00_thermodynamics")
        if data_g is not None:
            data_gibbs[measure].append((temp, data_g[:, 2], data_g[:, 3]))  # Media e errore

    # h = 0.02: Magnetizzazione
    measure = "magnetization"
    data_m = read_data(base_dir_metropolis, measure, temp, "h0.02_magnetization")
    if data_m is not None:
        data_metropolis[measure].append((temp, data_m[:, 2], data_m[:, 3]))  # Media e errore

    data_g = read_data(base_dir_gibbs, measure, temp, "h0.02_magnetization")
    if data_g is not None:
        data_gibbs[measure].append((temp, data_g[:, 2], data_g[:, 3]))  # Media e errore

# Funzione per plottare i dati su un axes specifico
def plot_measure_on_ax(ax, data_metropolis, data_gibbs, measure, ylabel, title, theoretical_curve=None):
    """
    Plot misure su un axes specifico.
    
    Parameters:
    -----------
    ax : matplotlib.axes.Axes
        L'axes su cui plottare
    data_metropolis : dict
        Dati Metropolis
    data_gibbs : dict
        Dati Gibbs
    measure : str
        Nome della misura
    ylabel : str
        Etichetta asse y
    title : str
        Titolo del subplot
    theoretical_curve : tuple or None
        Tupla (temps_theory, values_theory, label_theory) per la curva teorica, oppure None
    """
    temps_m = [d[0] for d in data_metropolis[measure]]
    means_m = [d[1][-1] for d in data_metropolis[measure]]  # Ultima media
    errors_m = [d[2][-1] for d in data_metropolis[measure]]  # Ultimo errore

    temps_g = [d[0] for d in data_gibbs[measure]]
    means_g = [d[1][-1] for d in data_gibbs[measure]]  # Ultima media
    errors_g = [d[2][-1] for d in data_gibbs[measure]]  # Ultimo errore

    # Plot dei dati numerici
    ax.errorbar(temps_m, means_m, yerr=errors_m, fmt='o', label="Metropolis", 
                color=color, markersize=7)
    ax.errorbar(temps_g, means_g, yerr=errors_g, fmt='s', label="Gibbs", 
                color=color2, markersize=7)
    
    # Plot della curva teorica se fornita
    if theoretical_curve is not None:
        temps_theory, values_theory, label_theory = theoretical_curve
        ax.plot(temps_theory, values_theory, '--', label=label_theory, 
                color="orange", linewidth=2.5, alpha=0.8)
    
    ax.set_xlabel("Temperature (T)", fontsize=11, fontweight='bold')
    ax.set_ylabel(ylabel, fontsize=11, fontweight='bold')
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.legend(fontsize=10, loc='best')
    ax.grid(True, linestyle="--", alpha=0.6)

# Crea la figura con 4 subplots (2x2)
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# ========================================================================
# PLOT 1: Total Energy vs Temperature
# ========================================================================
e = -J*( th + ch*thN )/( 1 + thN )
theoretical_energy = (T, e, "Ising Theory")
plot_measure_on_ax(axes[0, 0], data_metropolis, data_gibbs, 
                   "total_energy", "Total Energy", 
                   "Internal Energy (h=0)",
                   theoretical_energy)  # Aggiungi curva teorica qui

# ========================================================================
# PLOT 2: Specific Heat vs Temperature
# ========================================================================
heat=((b*J)**2)*(((1+thN+(Ns-1)*(th**2)+(Ns-1)*(ch**2)*thN)/(1+thN))-Ns*((th+ch*thN)/(1+thN))**2)
theoretical_specific_heat = (T, heat, "Ising Theory")
plot_measure_on_ax(axes[0, 1], data_metropolis, data_gibbs, 
                   "specific_heat", "Specific Heat (C/N)", 
                   "Specific Heat (h=0)",
                   theoretical_specific_heat)  # Aggiungi curva teorica qui

# ========================================================================
# PLOT 3: Susceptibility vs Temperature
# ========================================================================
X = b*np.exp(2*b*J)*(1-thN)/(1+thN)
theoretical_susceptibility = (T, X, "Ising Theory")
plot_measure_on_ax(axes[1, 0], data_metropolis, data_gibbs, 
                   "susceptibility", "Susceptibility (χ/N)", 
                   "Susceptibility (h=0)",
                   theoretical_susceptibility)  # Aggiungi curva teorica qui

# ========================================================================
# PLOT 4: Magnetization vs Temperature (h=0.02)
# ========================================================================
l1 = np.exp(b*J)*np.cosh(b*h)+np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))
l2 = np.exp(b*J)*np.cosh(b*h)-np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))
Z = l1**Ns + l2**Ns
M = (np.exp(b*J)*np.sinh(b*h)*((l1**(Ns-1))*(1+np.exp(b*J)*np.cosh(b*h)/np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))) 
        + (l2**(Ns-1))*(1-np.exp(b*J)*np.cosh(b*h)/np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J)))))/(Z)
theoretical_magnetization = (T, M, "Ising Theory")
plot_measure_on_ax(axes[1, 1], data_metropolis, data_gibbs, 
                   "magnetization", "Magnetization (|M|/N)", 
                   "Magnetization (h=0.02)",
                   theoretical_magnetization)  # Aggiungi curva teorica qui

plt.tight_layout()
plt.show()
```
